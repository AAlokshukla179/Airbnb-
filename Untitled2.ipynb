```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/Netflix Dataset.csv")
df.head()
```

```python
print("Nulls present in the dataset column wise")
df.isnull().sum()
```

```python
df.info()
```

```python
df.columns = [c.strip() for c in df.columns]
```

```python
import re
def extract_year(x):
    if pd.isna(x):
        return None
    s = str(x)
    m = re.search(r'(19|20)\d{2}', s)
    return int(m.group(0)) if m else None
df["Year"] = df["Release_Date"].apply(extract_year)
```

```python
print("Nulls present in the dataset column wise")
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)

#Drop all records with missing values
df.dropna(inplace=True)
```

```python
print("Dataset Shape:", df.shape)
print("\nMissing Values:\n", df.isna().sum())
print("\nSample Data:\n", df.head())
```

```python
df['Release_Date'] = pd.to_datetime(df['Release_Date'], format='%Y', errors='coerce')
print("\nNew dtypes:")
print(df.dtypes)
```

```python
def remove_outlier(col_name):
    sorted(col_name)
    Q1,Q3 = col_name.quantile([0.25,0.75])
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return lower_bound, upper_bound
```

```python
col_type = "Category"      # Movies/TV Shows stored here
col_title = "Title"
col_country = "Country"
col_release = "Release_Date"  # Has year info
col_genres = "Category"    # Also contains genres in this dataset
col_duration = "Duration"
col_rating = "Rating"
```

```python
# Standardize type column
df[col_type] = df[col_type].astype(str).str.strip().str.title()

# Extract release year
df["release_year"] = pd.to_numeric(
    df[col_release].astype(str).str.extract(r"(\d{4})")[0],
    errors="coerce"
).astype("Int64")
```

```python
# Genres: split comma-separated categories
df["genres_list"] = df[col_genres].fillna("Unknown").astype(str).apply(
    lambda x: [g.strip() for g in x.split(",") if g.strip() != ""]
)
df_genres = df.explode("genres_list").rename(columns={"genres_list": "genre"})
```

```python
# Primary country (first country if multiple listed)
df["primary_country"] = df[col_country].fillna("Unknown").astype(str).apply(
    lambda x: x.split(",")[0].strip() if str(x).strip() != "" else "Unknown"
)

# Year used for trend analysis (fallback to release_year)
df["added_year"] = df["release_year"]
```

```python
import os
outdir = "netflix_analysis_outputs"
os.makedirs(outdir, exist_ok=True)
```

 Movies vs TV Shows count

```python
counts = df[col_type].value_counts()
plt.figure(figsize=(6,4))
counts.plot(kind="bar", color=["blue", "orange"])
plt.title("Count by Type (Movies vs TV Shows)")
plt.xlabel("Type")
plt.ylabel("Count")
plt.savefig(os.path.join(outdir, "type_count.png"))
plt.show()
```

Top 15 Genres in TV Shows and Movie

```python
top_genres = df_genres["genre"].value_counts().head(15)
plt.figure(figsize=(10,5))
top_genres.plot(kind="bar", color="purple")
plt.title("Top 15 Genres")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.savefig(os.path.join(outdir, "top_genres.png"))
plt.show()
```

Genres split by Type(Movie or TV Shows)

```python
topn = 8
topn_genres = df_genres["genre"].value_counts().head(topn).index.tolist()
pivot = df_genres[df_genres["genre"].isin(topn_genres)].pivot_table(
    index="genre", columns=col_type, values=col_title, aggfunc="count", fill_value=0
)
pivot = pivot.loc[topn_genres]
pivot.plot(kind="bar", stacked=True, figsize=(10,6))
plt.title(f"Top {topn} Genres split by Type")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.savefig(os.path.join(outdir, "genres_by_type_stacked.png"))
plt.show()
```

Top 15 Countries contributing in viewership

```python
top_countries = df["primary_country"].value_counts().head(15)
plt.figure(figsize=(10,5))
top_countries.plot(kind="bar", color="brown")
plt.title("Top 15 Primary Countries")
plt.xlabel("Country")
plt.ylabel("Count")
plt.savefig(os.path.join(outdir, "top_countries.png"))
plt.show()
```
